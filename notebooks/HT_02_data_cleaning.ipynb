#  HT_02 — Data Cleaning

Clean the raw dataset: `?` → NaN → drop, coerce numerics, remove duplicates.

*Related: [Pipeline](../docs/pipeline.md)*


```python
# Repo root on sys.path so `src` imports work
import sys
from pathlib import Path
sys.path.insert(0, str(Path.cwd().parent))

import pandas as pd
from src.data.cleaner import clean
from src.data.loader import load_raw
```

```python
raw = load_raw('../data/raw/cleve.mod')
print(f'Raw: {len(raw)} rows')
raw.head()
```

## 1. Missing Values Before Cleaning


```python
(raw == '?').sum()[lambda s: s > 0]
```

```python
rows_with_missing = raw.isin(['?']).any(axis=1).sum()
print(f'Rows containing at least one missing value: {rows_with_missing}')
```

## 2. Apply Cleaning Pipeline


```python
df = clean(raw)
print(f'Clean: {len(df)} rows')
```

```python
assert df.isna().sum().sum() == 0, 'nulls remain'
assert not df.isin(['?']).any().any(), 'missing markers remain'
print('No missing values or ? markers remain.')
```

## 3. Duplicates Check


```python
dupes = raw.drop_duplicates(keep=False).shape[0] - df.shape[0]
print(f'Duplicate rows removed during cleaning: {len(raw) - rows_with_missing - len(df) + 0}')
print(f'Rows lost total: {len(raw) - len(df)}')
```

## 4. Save Interim Output


```python
df.to_csv('../data/interim/cleve_interim.csv', index=False)
print('saved -> data/interim/cleve_interim.csv')
```
