## 3. Missing Values

Missing values are encoded as `?` (expected in `ca` and `thal`).


```python
(df == '?').sum()[lambda s: s > 0]
```

## 4. Numeric Distributions


```python
import matplotlib.pyplot as plt
import seaborn as sns

num_cols = ['age', 'trestbps', 'chol', 'thalach', 'oldpeak']
df_num = df[num_cols].apply(pd.to_numeric, errors='coerce')

fig, axes = plt.subplots(2, 3, figsize=(15, 8))
for ax, col in zip(axes.flat, num_cols):
    sns.histplot(df_num[col].dropna(), kde=True, ax=ax, color='#C0392B')
    ax.set_title(col)
axes.flat[-1].axis('off')
plt.tight_layout()
plt.show()
```

## 5. Outliers

Known outliers: `chol` up to 564 mg/dl and `oldpeak` up to 6.2.


```python
df_num.agg(['min', 'max']).T
```

## Key Findings

- 303 rows, 15 columns (mixed numeric/symbolic)
- ±164 `buff` vs ±139 `sick` — relatively balanced
- `?` missing values in `ca` and `thal` (±6 rows)
- Outliers in `chol` (564) and `oldpeak` (6.2)
